 #Task4:نتائج الاستكشاف والتحليل الإحصائي

```python
rows, cols = df.shape
print(f" عدد الصفوف: {rows}, عدد الأعمدة: {cols}")
```

```python
print("\n عدد القيم المفقودة لكل عمود:")
print(df.isnull().sum())
```

```python
duplicates_count = df.duplicated().sum()
print(f"\n عدد الصفوف المكررة: {duplicates_count}")
```

```python
unique_species = df['species'].nunique()
print(f"\n عدد الأنواع الفريدة: {unique_species}")
```

```python
print("\n عدد الأزهار لكل نوع:")
print(df['species'].value_counts())
```

```python
mean_sepal_length = df['sepal_length'].mean()
print(f"\n متوسط طول السبلة: {mean_sepal_length:.2f}")
```

```python
max_petal_length = df['petal_length'].max()
print(f" أقصى طول للبتلة: {max_petal_length}")
```

```python
min_sepal_width = df['sepal_width'].min()
print(f" أقل عرض للسبلة: {min_sepal_width}")
```

```python
print("\n متوسط الأعمدة الرقمية لكل نوع زهرة:")
display(df.groupby('species').mean())
```

```python
print("\n الإحصاءات الوصفية (Describe):")
display(df.describe())
```

#Task 5 : تصفية وتحليل البيانات

```python
print("--- الأزهار بطول سبلة أكبر من 6 ---")
display(df[df['sepal_length'] > 6].head())
```

```python
print("\n--- الأزهار بعرض بتلة أكبر من 1.5 ---")
display(df[df['petal_width'] > 1.5].head())
```

```python
print("\n--- أزهار نوع setosa فقط ---")
display(df[df['species'] == 'setosa'].head())
```

```python
mean_petal_by_species = df.groupby('species')['petal_length'].mean()
highest_avg_species = mean_petal_by_species.idxmax()
print(f"\n النوع الذي يمتلك أعلى متوسط طول بتلة هو: {highest_avg_species} بمتوسط {mean_petal_by_species.max():.2f}")
```

```python
print("\n--- ترتيب البيانات تنازلياً حسب طول السبلة ---")
display(df.sort_values(by='sepal_length', ascending=False).head())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(9, 5))
sns.scatterplot(
    data=df, 
    x='sepal_length', 
    y='petal_length', 
    hue='species', 
    style='species', 
    s=100, 
    palette='Set2'
)
plt.title('Relationship Between Sepal Length and Petal Length by Species', fontsize=12, fontweight='bold')
plt.xlabel('Sepal Length (cm)', fontsize=10)
plt.ylabel('Petal Length (cm)', fontsize=10)
plt.legend(title='Species')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
